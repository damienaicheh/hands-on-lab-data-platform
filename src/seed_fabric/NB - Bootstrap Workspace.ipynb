## Setup &amp; helpers

```python
import json, time
import sempy.fabric as fabric
import sempy_labs as labs
import sempy_labs.lakehouse as lake
import sempy_labs.report as rep
import notebookutils


def current_workspace_id():
    return notebookutils.runtime.context["currentWorkspaceId"]


def item_exists(name, item_type, workspace=None):
    df = fabric.list_items(workspace=workspace, type=item_type)
    col = "Display Name" if "Display Name" in df.columns else df.columns[0]
    return name in set(df[col].astype(str))


def _artifact_id(artifact):
    if isinstance(artifact, dict):
        return artifact.get("id") or artifact.get("Id")
    return getattr(artifact, "id", None)
```

## 3. Semantic Model (import mode)

```python
# Re-point the Direct Lake model's OneLake connection to the NEW lakehouse.
# The BIM expression references a hard-coded OneLake path:
#   https://onelake.dfs.fabric.microsoft.com/{workspaceId}/{lakehouseId}
# Replace every occurrence of the old path with one built from the target
# workspace id and the newly created lakehouse id.
OLD_ONELAKE_URL = "https://onelake.dfs.fabric.microsoft.com/7a35cc4f-f866-4a81-b753-a6c63e722407/c9f8d73d-c9d4-442a-a1e4-3ba0a5f881c8"

def rewrite_connection(bim, old_url, new_url):
    text = json.dumps(bim)
    if old_url and new_url:
        text = text.replace(old_url, new_url)
    return json.loads(text)

NEW_ONELAKE_URL = f"https://onelake.dfs.fabric.microsoft.com/{ws_id}/{lakehouse_id}"

model_bim = rewrite_connection(model_bim, OLD_ONELAKE_URL, NEW_ONELAKE_URL)
print("✓ Rewrote model OneLake connection to the new lakehouse.")
print("  New OneLake URL:", NEW_ONELAKE_URL)
```

```python
import base64
import copy
import json
from sempy_labs._helper_functions import lro

if item_exists(REPORT_NAME, "Report", TARGET_WORKSPACE):
    print(f"✓ Report '{REPORT_NAME}' already exists — skipping create.")
else:
    # The exported PBIR references the semantic model with a relative 'byPath'
    # reference. The Fabric REST API only supports 'byConnection', so rewrite
    # definition.pbir to point at the semantic model via an XMLA connection.
    sm_id = next(
        i["id"]
        for i in client.get(f"/v1/workspaces/{ws_id}/semanticModels").json()["value"]
        if i["displayName"] == DATASET_NAME
    )
    ws_name = fabric.resolve_workspace_name(ws_id)

    print(sm_id)

    pbir = {  
        "$schema": "https://developer.microsoft.com/json-schemas/fabric/item/report/definitionProperties/2.0.0/schema.json",
        "version": "4.0",
        "datasetReference": {
            "byConnection": {          
                "connectionString": f"semanticModelId={sm_id}"
            }
        }
    }

    report_payload = copy.deepcopy(report_json)
    pbir_payload = base64.b64encode(json.dumps(pbir).encode("utf-8")).decode("utf-8")
    for part in report_payload["definition"]["parts"]:
        if part["path"] == "definition.pbir":
            part["payload"] = pbir_payload
            break
    else:
        # definition.pbir was not present — add it so the report is bound.
        report_payload["definition"]["parts"].append({
            "path": "definition.pbir",
            "payload": pbir_payload,
            "payloadType": "InlineBase64",
        })

    # Create the report item from its PBIR definition.
    lro(
        client,
        client.post(
            f"/v1/workspaces/{ws_id}/reports",
            json=report_payload,
        ),
        status_codes=[201, 202],
    )
    print(f"✓ Created report '{REPORT_NAME}' bound to '{DATASET_NAME}'.")
```
